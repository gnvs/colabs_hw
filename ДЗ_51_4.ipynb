```python
%%writefile src/utils/notifications.py

import os
import asyncio
from dotenv import load_dotenv
from aiogram import Bot

load_dotenv()

TELEGRAM_BOT_TOKEN = os.getenv("TELEGRAM_BOT_TOKEN")
TELEGRAM_CHAT_ID = os.getenv("TELEGRAM_CHAT_ID")
LOW_BALANCE_THRESHOLD = float(
    os.getenv("LOW_BALANCE_THRESHOLD", "10")
)


async def send_telegram_notification(message: str):
    """Отправляет уведомление администратору в Telegram."""

    if not TELEGRAM_BOT_TOKEN:
        print("Ошибка: не указан TELEGRAM_BOT_TOKEN")
        return False

    if not TELEGRAM_CHAT_ID:
        print("Ошибка: не указан TELEGRAM_CHAT_ID")
        return False

    bot = Bot(token=TELEGRAM_BOT_TOKEN)

    try:
        await bot.send_message(
            chat_id=TELEGRAM_CHAT_ID,
            text=message
        )

        print("Telegram-уведомление отправлено")
        return True

    except Exception as e:
        print(f"Ошибка отправки Telegram: {e}")
        return False

    finally:
        await bot.session.close()


def check_balance_and_notify(balance: float):
    """
    Проверяет баланс и отправляет уведомление,
    если он ниже установленного порога.
    """

    if balance < LOW_BALANCE_THRESHOLD:

        message = (
            "ВНИМАНИЕ!\n\n"
            "Обнаружен низкий баланс приложения.\n\n"
            f"Текущий баланс: {balance:.2f} ₽\n"
            f"Минимальный баланс: "
            f"{LOW_BALANCE_THRESHOLD:.2f} ₽"
        )

        return asyncio.run(
            send_telegram_notification(message)
        )

    print(
        f"Баланс в норме: "
        f"{balance:.2f} ₽"
    )

    return False
```
