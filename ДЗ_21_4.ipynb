```python

!pip -q install navec

# Стандартные библиотеки
import os
import random
import shutil
import zipfile
from collections import defaultdict
from pathlib import Path, PurePosixPath

# Библиотеки для вычислений и визуализации
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf

# Navec — предобученные представления слов проекта Natasha
from navec import Navec

# Метрики для итоговой оценки модели на тестовой выборке
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
)

# Колбэки для сохранения лучшей модели и управления обучением
from tensorflow.keras.callbacks import (
    ModelCheckpoint,
    EarlyStopping,
    ReduceLROnPlateau,
)

# Слои
from tensorflow.keras.layers import (
    Input,
    Embedding,
    SpatialDropout1D,
    BatchNormalization,
    Bidirectional,
    LSTM,
    GRU,
    Dropout,
    Dense,
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical


# Фиксируем seed, чтобы результаты между запусками были более воспроизводимыми
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)

print("TensorFlow:", tf.__version__)

# Если список пустой, обучение пойдет на CPU
print("GPU:", tf.config.list_physical_devices("GPU"))



# Максимальный размер словаря
MAX_WORDS = 15_000

# В названии файла Navec есть 300d, поэтому Embedding будет 300
EMBEDDING_DIM = 300

# Длина одного текстового фрагмента
# Один обучающий пример будет состоять из 1000 токенов
WIN_SIZE = 1000

# Шаг скользящего окна
# Чем меньше шаг, тем больше обучающих примеров и тем сильнее окна перекрываются
WIN_STEP = 100

# По 10% текста каждого автора выделяем на validation и test
# Оставшиеся 80% идут в train
VAL_SPLIT = 0.10
TEST_SPLIT = 0.10

# Балансировка классов:
# берем одинаковое число токенов у каждого писателя, чтобы крупный класс не доминировал над остальными
MAX_BALANCE_TOKENS = 260_000

# Параметры обучения
BATCH_SIZE = 32
EPOCHS = 35

# Используем 4 авторов
# Выбраны авторы с достаточно большими объемами текста
CLASS_LIST = [
    "Dostoevsky",
    "Tolstoy",
    "Gogol",
    "Turgenev",
]

# Ссылки на датасет и предобученные Navec-вектора
DATASET_URL = (
    "https://storage.yandexcloud.net/academy.ai/"
    "russian_literature.zip"
)

NAVEC_URL = (
    "https://storage.yandexcloud.net/natasha-navec/packs/"
    "navec_hudlit_v1_12B_500K_300d_100q.tar"
)

# Рабочая директория и файл
DATASET_DIR = Path("/content/russian_literature")
BEST_WEIGHTS = "/content/best_writers.weights.h5"


# Скачиваем архив с произведениями.
# tf.keras.utils.get_file использует кеш, поэтому повторно файл обычно не скачивается
dataset_zip = tf.keras.utils.get_file(
    "russian_literature.zip",
    origin=DATASET_URL,
)

# Скачиваем предобученные вектора Natasha / Navec.
navec_path = tf.keras.utils.get_file(
    "navec_hudlit_v1_12B_500K_300d_100q.tar",
    origin=NAVEC_URL,
)


def repair_zip_name(name):
    """
    В ZIP встречаются русские имена файлов.
    Иногда zipfile читает UTF-8 имя как CP437, и появляются строки как абракадабра
    Пробуем восстановить нормальное UTF-8 имя.
    """
    try:
        return name.encode("cp437").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return name


def shorten_component(component, max_bytes=220):
    """
    Linux ограничивает длину одного компонента пути примерно 255 байтами.
    Некоторые названия произведений очень длинные, поэтому при необходимости аккуратно сокращаем только имя файла.
    """
    if len(component.encode("utf-8")) <= max_bytes:
        return component

    stem, suffix = os.path.splitext(component)
    allowed = max(
        20,
        max_bytes - len(suffix.encode("utf-8")),
    )

    result = ""

    for char in stem:
        # Добавляем символ, только если не выйдем за лимит
        if len((result + char).encode("utf-8")) > allowed:
            break
        result += char

    return result.rstrip() + suffix


def safe_extract_zip(zip_path, destination):
    """
    Распаковывает архив:
    исправляет кодировку русских имен
    сокращает слишком длинные имена
    проверяет, что файл не выйдет за пределы destination
    """
    destination = Path(destination)
    destination.mkdir(
        parents=True,
        exist_ok=True,
    )

    root = destination.resolve()

    with zipfile.ZipFile(zip_path, "r") as archive:

        for member in archive.infolist():

            # Восстанавливаем имя и приводим разделители пути к /
            name = repair_zip_name(
                member.filename
            ).replace("\\", "/")

            # Очищаем путь от опасных элементов . и ..
            parts = [
                shorten_component(part)
                for part in PurePosixPath(name).parts
                if part not in ("", ".", "..", "/")
            ]

            if not parts:
                continue

            target = destination.joinpath(*parts)
            resolved = target.resolve()

            # Защита от path traversal
            if resolved != root and root not in resolved.parents:
                raise RuntimeError(
                    "Небезопасный путь: " + member.filename
                )

            # Если это директория — просто создаем ее
            if member.is_dir():
                target.mkdir(
                    parents=True,
                    exist_ok=True,
                )
                continue

            # Создаем родительские каталоги файла
            target.parent.mkdir(
                parents=True,
                exist_ok=True,
            )

            # Копируем содержимое файла из архива
            with archive.open(member, "r") as src:
                with open(target, "wb") as dst:
                    shutil.copyfileobj(src, dst)


# Удаляем старую распакованную версию, чтобы не смешивать файлы разных запусков
if DATASET_DIR.exists():
    shutil.rmtree(DATASET_DIR)

safe_extract_zip(
    dataset_zip,
    DATASET_DIR,
)

print("Распаковано:", DATASET_DIR)
print(
    "Папки:",
    sorted(x.name for x in DATASET_DIR.iterdir()),
)

def read_text_safely(path):
    """
    Читаем текст произведения.
    В датасете могут встречаться файлы в UTF-8 и Windows-1251.
    """
    raw = Path(path).read_bytes()

    for encoding in ("utf-8", "cp1251"):
        try:
            return raw.decode(encoding)
        except UnicodeDecodeError:
            pass

    # Последний резервный вариант: игнорируем поврежденные символы
    return raw.decode(
        "utf-8",
        errors="ignore",
    )


def collect_author_files(root):
    """
    Собираем все txt-файлы каждого автора
    Автор может находиться как в prose, так и в poems
    """
    result = defaultdict(list)

    for genre in ("prose", "poems"):

        genre_dir = Path(root) / genre

        if not genre_dir.exists():
            continue

        for author_dir in genre_dir.iterdir():

            if author_dir.is_dir():
                result[author_dir.name].extend(
                    sorted(
                        author_dir.glob("*.txt")
                    )
                )

    return dict(result)


# Получаем словарь:
# имя автора -> список его текстовых файлов
author_files = collect_author_files(
    DATASET_DIR
)

# Проверяем, что выбранные нами авторы действительно есть в архиве
missing = [
    author
    for author in CLASS_LIST
    if author not in author_files
]

if missing:
    raise RuntimeError(
        "Не найдены авторы: "
        + ", ".join(missing)
    )


# Объединяем все произведения каждого автора в один большой текст.
# В all_texts позиция текста соответствует позиции автора в CLASS_LIST.
all_texts = []

for author in CLASS_LIST:

    text = "\n".join(
        read_text_safely(path)
        for path in author_files[author]
    )

    all_texts.append(text)

    print(
        f"{author:15s}: "
        f"{len(author_files[author]):3d} файлов, "
        f"{len(text):,} символов"
    )


# Создаем токенизатор.
# num_words ограничивает словарь 15000 наиболее частыми словами.
tokenizer = Tokenizer(
    num_words=MAX_WORDS,

    # Удаляем пунктуацию и служебные символы
    filters='!"#$%&()*+,-–—./…:;<=>?@[\\]^_`{|}~«»\t\n\xa0\ufeff',

    # Приводим все слова к нижнему регистру
    lower=True,

    # Разделитель слов
    split=" ",

    # Слова вне словаря заменяются специальным токеном
    oov_token="неизвестное_слово",

    char_level=False,
)

# Строим частотный словарь на текстах выбранных авторов
tokenizer.fit_on_texts(
    all_texts
)

# Каждый текст преобразуем в последовательность целочисленных индексов слов.
# Типа того: ["я", "иду"] -> [25, 137]
seq_list = tokenizer.texts_to_sequences(
    all_texts
)

# Смотрим, сколько токенов получилось у каждого автора
for author, sequence in zip(
    CLASS_LIST,
    seq_list,
):
    print(
        f"{author:15s}: "
        f"{len(sequence):,} токенов"
    )

# Находим размер самого маленького класса
# Нельзя взять у остальных авторов больше токенов
# иначе классы снова будут несбалансированными
min_tokens = min(
    len(sequence)
    for sequence in seq_list
)

# Ограничиваем объем каждого автора
# В результате каждый класс содержит одинаковое число токенов
balance_tokens = min(
    min_tokens,
    MAX_BALANCE_TOKENS,
)

seq_balance = [
    sequence[:balance_tokens]
    for sequence in seq_list
]

print(
    "Токенов на автора:",
    balance_tokens,
)

# Проверяем балансировку
for author, sequence in zip(
    CLASS_LIST,
    seq_balance,
):
    print(
        author,
        len(sequence),
    )


# ===

def seq_split(sequence, win_size, step):
    """
    Нарезает последовательность токенов скользящим окном.

    Пример:
    sequence = [1,2,3,4,5,6]
    win_size = 3
    step = 2

    Получим:
    [1,2,3], [3,4,5]
    """
    return [
        sequence[i:i + win_size]
        for i in range(
            0,
            len(sequence) - win_size + 1,
            step,
        )
    ]


def seq_vectorize(
    seq_list,
    val_split,
    test_split,
    class_list,
    win_size,
    step,
):
    """
    Модифицированная функция из практического занятия.

    Исходная функция возвращала только train и test.
    По условию домашней работы она должна вернуть 3 выборки:

    x_train, y_train
    x_val,   y_val
    x_test,  y_test

    ВАЖНО:
    сначала последовательность каждого автора делится на 80% / 10% / 10%,
    и только потом каждая часть режется на окна.

    Благодаря этому одно и то же пересекающееся окно не может одновременно
    попасть, например, и в train, и в test.
    """

    # Здесь будем накапливать признаки X и метки Y
    x_train, y_train = [], []
    x_val, y_val = [], []
    x_test, y_test = [], []

    # Проверяем корректность долей
    if val_split + test_split >= 1:
        raise ValueError(
            "val_split + test_split должно быть < 1"
        )

    # Обрабатываем каждого писателя отдельно
    for cls, sequence in enumerate(seq_list):

        # Полная длина последовательности токенов текущего автора
        sequence_length = len(sequence)

        # Граница train:
        # 1 - 0.10 validation - 0.10 test = 0.80
        train_gate = int(
            sequence_length
            * (1 - val_split - test_split)
        )

        # Граница validation:
        # все, что после нее, пойдет в test
        val_gate = int(
            sequence_length
            * (1 - test_split)
        )

        # Разделяем текст автора на три НЕПЕРЕСЕКАЮЩИЕСЯ по исходному тексту части
        train_sequence = sequence[:train_gate]
        val_sequence = sequence[
            train_gate:val_gate
        ]
        test_sequence = sequence[val_gate:]

        # Только теперь нарезаем каждую часть скользящими окнами
        vectors_train = seq_split(
            train_sequence,
            win_size,
            step,
        )

        vectors_val = seq_split(
            val_sequence,
            win_size,
            step,
        )

        vectors_test = seq_split(
            test_sequence,
            win_size,
            step,
        )

        # Преобразуем номер класса в one-hot метку.
        # Например для 4 авторов класс 2 -> [0, 0, 1, 0]
        label = to_categorical(
            cls,
            len(class_list),
        )

        # Добавляем окна и соответствующие им метки в train
        x_train += vectors_train
        y_train += (
            [label] * len(vectors_train)
        )

        # Добавляем окна и метки в validation
        x_val += vectors_val
        y_val += (
            [label] * len(vectors_val)
        )

        # Добавляем окна и метки в test
        x_test += vectors_test
        y_test += (
            [label] * len(vectors_test)
        )

    # Keras работаеи с numpy-массивами, поэтому преобразуем все шесть списков перед возвратом.
    return (
        np.asarray(
            x_train,
            dtype=np.int32,
        ),
        np.asarray(
            y_train,
            dtype=np.float32,
        ),
        np.asarray(
            x_val,
            dtype=np.int32,
        ),
        np.asarray(
            y_val,
            dtype=np.float32,
        ),
        np.asarray(
            x_test,
            dtype=np.int32,
        ),
        np.asarray(
            y_test,
            dtype=np.float32,
        ),
    )


# Формируем все три выборки
(
    x_train,
    y_train,
    x_val,
    y_val,
    x_test,
    y_test,
) = seq_vectorize(
    seq_balance,
    VAL_SPLIT,
    TEST_SPLIT,
    CLASS_LIST,
    WIN_SIZE,
    WIN_STEP,
)

# Проверяем размеры.
# В каждой строке X находится одно окно длиной WIN_SIZE токенов.
print(
    "TRAIN:",
    x_train.shape,
    y_train.shape,
)
print(
    "VAL:  ",
    x_val.shape,
    y_val.shape,
)
print(
    "TEST: ",
    x_test.shape,
    y_test.shape,
)


# ===
def show_balance(y, title):
    """
    Показывает число окон каждого класса.
    После балансировки значения для всех авторов должны быть одинаковыми.
    """

    # one-hot -> номер класса
    labels = np.argmax(
        y,
        axis=1,
    )

    counts = np.bincount(
        labels,
        minlength=len(CLASS_LIST),
    )

    table = pd.DataFrame(
        {
            "Автор": CLASS_LIST,
            "Примеров": counts,
        }
    )

    display(
        table.style.set_caption(title)
    )


# Проверяем баланс отдельно для каждой выборки
show_balance(
    y_train,
    "TRAIN",
)
show_balance(
    y_val,
    "VALIDATION",
)
show_balance(
    y_test,
    "TEST",
)


# ===
# Загружаем предобученную модель Navec.
# Это готовые 300-мерные векторы русских слов, обученные заранее на большом корпусе художественной литературы.
navec = Navec.load(
    navec_path
)

# Создаем матрицу весов Embedding.
# Строка с индексом i должна содержать Navec-вектор слова, которому tokenizer присвоил индекс i.
embedding_matrix = np.zeros(
    (
        MAX_WORDS,
        EMBEDDING_DIM,
    ),
    dtype=np.float32,
)

found = 0

# tokenizer.word_index:
# слово -> его целочисленный индекс
for word, index in tokenizer.word_index.items():

    # В модели используем только первые MAX_WORDS слов
    if index >= MAX_WORDS:
        continue

    # Ищем готовый вектор данного слова в Navec
    vector = navec.get(word)

    if vector is not None:

        # Записываем вектор точно в ту строку,
        # которая соответствует индексу слова в Keras Tokenizer
        embedding_matrix[index] = vector

        found += 1


print(
    "Размер embedding_matrix:",
    embedding_matrix.shape,
)

print(
    "Найдено слов в Navec:",
    found,
)


# ===
# Строим лучшую архитектуру из практического занятия.
model = Sequential(
    [
        # На вход подается окно из WIN_SIZE целочисленных индексов слов
        Input(
            shape=(WIN_SIZE,),
            dtype="int32",
        ),

        # Embedding переводит каждый индекс слова в вектор длиной 300.
        # weights=[embedding_matrix] загружает предобученные веса Natasha / Navec.

        # trainable=False замораживает Embedding.
        # Во время обучения Navec-вектора изменяться не будут.
        Embedding(
            input_dim=MAX_WORDS,
            output_dim=EMBEDDING_DIM,
            weights=[embedding_matrix],
            trainable=False,
            name="navec_embedding",
        ),

        # SpatialDropout1D выключает целые векторы отдельных слов, а не отдельные координаты внутри вектора.
        # уменьшает переобучение.
        SpatialDropout1D(0.3),

        BatchNormalization(),

        # Двунаправленный LSTM читает последовательность
        # одновременно слева направо и справа налево.
        #
        # return_sequences=True нужен, потому что дальше идет еще один рекуррентный слой, которому нужна последовательность состояний.
        Bidirectional(
            LSTM(
                8,
                return_sequences=True,
            )
        ),

        # Второй двунаправленный LSTM
        Bidirectional(
            LSTM(
                8,
                return_sequences=True,
            )
        ),

        Dropout(0.3),
        BatchNormalization(),

        # Первый GRU тоже должен вернуть всю последовательность, потому что после него расположен еще один GRU.
        GRU(
            16,
            return_sequences=True,
            reset_after=True,
        ),

        # Последний GRU возвращает уже один итоговый вектор признаков
        GRU(
            16,
            reset_after=True,
        ),

        Dropout(0.3),
        BatchNormalization(),

        # Полносвязный слой объединяет признаки, выделенные рекуррентными слоями
        Dense(
            100,
            activation="relu",
        ),

        Dropout(0.3),
        BatchNormalization(),

        # Финальный слой имеет столько нейронов, сколько писателей мы классифицируем.
        # softmax дает распределение вероятностей по авторам.
        Dense(
            len(CLASS_LIST),
            activation="softmax",
        ),
    ]
)


# categorical_crossentropy используется, потому что метки y представлены в one-hot формате.
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3,
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)

model.summary()

# Дополнительная явная проверка требования: Embedding действительно заморожен
print(
    "Embedding trainable:",
    model.get_layer(
        "navec_embedding"
    ).trainable,
)

assert (
    model.get_layer(
        "navec_embedding"
    ).trainable
    is False
)


# ====
# ModelCheckpoint сохраняет веса только той эпохи, на которой validation accuracy была максимальной.
checkpoint = ModelCheckpoint(
    BEST_WEIGHTS,
    monitor="val_accuracy",
    mode="max",
    save_best_only=True,
    save_weights_only=True,
    verbose=1,
)

# Если validation accuracy долго не улучшается, останавливаем обучение, чтобы не переобучать модель.
early = EarlyStopping(
    monitor="val_accuracy",
    mode="max",
    patience=7,
    verbose=1,
)

# Если val_loss перестал уменьшаться, постепенно снижаем learning rate.
reduce = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    min_lr=1e-6,
    verbose=1,
)


history = model.fit(
    # Обучающая выборка используется для изменения весов модели
    x_train,
    y_train,

    # validation передается явно через validation_data,
    # а не создается автоматически через validation_split.
    validation_data=(
        x_val,
        y_val,
    ),

    epochs=EPOCHS,
    batch_size=BATCH_SIZE,

    # Перемешиваем обучающие окна между эпохами
    shuffle=True,

    callbacks=[
        checkpoint,
        early,
        reduce,
    ],

    verbose=1,
)


# ===
epochs=range(1,len(history.history['loss'])+1)
plt.figure(figsize=(12,5)); plt.plot(epochs,history.history['accuracy'],label='train'); plt.plot(epochs,history.history['val_accuracy'],label='validation'); plt.xlabel('Эпоха'); plt.ylabel('Accuracy'); plt.grid(); plt.legend(); plt.show()
plt.figure(figsize=(12,5)); plt.plot(epochs,history.history['loss'],label='train'); plt.plot(epochs,history.history['val_loss'],label='validation'); plt.xlabel('Эпоха'); plt.ylabel('Loss'); plt.grid(); plt.legend(); plt.show()

# ===
# Загружаем веса ЛУЧШЕЙ эпохи, сохраненные ModelCheckpoint по максимальному val_accuracy.
model.load_weights(
    BEST_WEIGHTS
)

# Обычная accuracy на тестовой выборке.
# Test до этого не участвовал ни в обучении, ни в выборе лучшей эпохи.
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    batch_size=BATCH_SIZE,
    verbose=0,
)

# Получаем распределение вероятностей для каждого тестового окна.
probabilities = model.predict(
    x_test,
    batch_size=BATCH_SIZE,
    verbose=0,
)

# One-hot метки превращаем обратно в номера классов
y_true = np.argmax(
    y_test,
    axis=1,
)

# Для прогноза выбираем класс с максимальной вероятностью
y_pred = np.argmax(
    probabilities,
    axis=1,
)


# Матрица ошибок строится именно по TEST, а не по validation.
#
# normalize="true" нормирует каждую строку:
# диагональный элемент показывает долю правильно распознанных примеров соответствующего автора.
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(
        len(CLASS_LIST)
    ),
    normalize="true",
)

cm = np.around(
    cm,
    3,
)


# Берем диагональ нормализованной матрицы ошибок и усредняем точность по всем писателям
mean_accuracy = np.diag(
    cm
).mean()


print(
    f"Test accuracy: "
    f"{test_accuracy:.2%}"
)

print(
    "Средняя точность по диагонали: "
    f"{mean_accuracy:.2%}"
)


# Проверяем прохождение порогов задания
if mean_accuracy >= 0.90:
    print(
        "Условие >90% выполнено."
    )

elif mean_accuracy >= 0.70:
    print(
        "Условие >70% выполнено."
    )

else:
    print(
        "Ниже 70% — "
        "смотрите рекомендации в конце."
    )


# Визуализируем нормализованную матрицу ошибок
fig, ax = plt.subplots(
    figsize=(9, 8)
)

ConfusionMatrixDisplay(
    cm,
    display_labels=CLASS_LIST,
).plot(
    ax=ax,
    values_format=".2f",
    colorbar=False,
)

plt.title(
    "Матрица ошибок на TEST"
)

plt.xticks(
    rotation=45,
    ha="right",
)

plt.tight_layout()
plt.show()


# Дополнительно выводим precision / recall / f1-score по каждому автору
print(
    classification_report(
        y_true,
        y_pred,
        target_names=CLASS_LIST,
        digits=3,
    )
)


# ===
# Текст генерируем на deepseek
MY_TEXT = """
Утро начиналось спокойно, но эта тишина почему-то не приносила облегчения.
Я вышел на улицу раньше обычного и долго шел без определенной цели.
Город уже просыпался: открывались маленькие магазины, во дворах слышались
голоса, а на перекрестке торопливо проходили люди, каждый занятый своей
мыслью. Мне же казалось странным, что среди этого привычного движения
можно чувствовать себя совершенно отдельно от всех.

Накануне я долго думал о том, почему человек так часто откладывает важные
решения. Мы убеждаем себя, что завтра обстоятельства станут понятнее,
что появится свободное время или нужная смелость. Но наступает завтра,
и вместе с ним приходят новые причины ничего не менять.

У старого дома я остановился. На втором этаже было открыто окно, занавеска
медленно двигалась от ветра. Когда-то я часто проходил здесь и не замечал
этого дома. Теперь же он почему-то напомнил мне о людях, с которыми мы
расстаемся без ссор и громких слов.

Я пошел дальше и неожиданно почувствовал, что тревога стала слабее.
Ничего вокруг не изменилось: тот же серый тротуар, те же машины и редкое
августовское солнце. Изменилось лишь отношение к собственным мыслям.
Иногда решение не приходит в виде точного ответа. Иногда достаточно
перестать от него убегать.
"""


# ===
def prepare_custom_text(
    text,
    tokenizer,
    win_size,
):
    """
    Подготавливает произвольный текст к той же форме,
    в которой модель видела данные при обучении.
    """

    # Используем ТОТ ЖЕ tokenizer, что и для обучающего датасета.
    # Иначе индексы слов не будут соответствовать Embedding.
    sequence = tokenizer.texts_to_sequences(
        [text]
    )[0]

    if not sequence:
        raise ValueError(
            "После токенизации текст пуст"
        )

    # Если пользовательский текст короче WIN_SIZE,
    # дополняем его нулями справа до нужной длины.
    if len(sequence) < win_size:
        return pad_sequences(
            [sequence],
            maxlen=win_size,
            padding="post",
            truncating="post",
        ).astype(np.int32)

    # Если текст длинный, делим его на несколько окон
    # Используем перекрытие 50%, чтобы охватить весь текст
    step = max(
        win_size // 2,
        1,
    )

    windows = seq_split(
        sequence,
        win_size,
        step,
    )

    # Добавляем последнее окно с конца текста
    # если оно еще не попало в список
    last_window = sequence[
        -win_size:
    ]

    if (
        not windows
        or windows[-1] != last_window
    ):
        windows.append(
            last_window
        )

    return np.asarray(
        windows,
        dtype=np.int32,
    )


# Подготавливаем собственный текст
my_windows = prepare_custom_text(
    MY_TEXT,
    tokenizer,
    WIN_SIZE,
)

# Модель выдает вероятность каждого автора для каждого окна текста
# Если окон несколько — усредняем вероятности
my_probabilities = model.predict(
    my_windows,
    verbose=0,
).mean(
    axis=0
)


# Для удобства - вывод вероятности таблицей
result = pd.DataFrame(
    {
        "Автор": CLASS_LIST,
        "Вероятность": my_probabilities,
    }
).sort_values(
    "Вероятность",
    ascending=False,
).reset_index(
    drop=True
)

display(result)


# Автор с максимальной вероятностью — главный ответ модели
print(
    "Больше всего текст похож на: "
    f"{result.loc[0, 'Автор']} "
    f"({result.loc[0, 'Вероятность']:.2%})"
)


# ===
# Круговая диаграмма показывает,
# какую вероятность модель присвоила каждому из четырех писателей.
plt.figure(
    figsize=(9, 9)
)

plt.pie(
    my_probabilities,
    labels=CLASS_LIST,
    autopct="%1.1f%%",
    startangle=90,
)

plt.title(
    "Вероятности авторства пользовательского текста"
)

# equal делает круг действительно круглым, а не овальным
plt.axis("equal")

plt.show()

```
