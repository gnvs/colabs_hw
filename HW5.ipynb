**Условие** Используя примеры из теоретической части 5.1 и 5.2, составьте класс Neuron модели нейрона, принимающего на вход от одного до 10 входных параметров. Запрещается пользоваться библиотекой numpy. На вход подается список. Матричное умножение реализуется с помощью циклов самостоятельно. Класс содержит методы для вычисления взвешенной суммы, а также содержит функции активации из урока.

```python
class Neuron:
    def __init__(self, w):
        # Список весовых коэффициентов (от 1 до 10 элементов)
        if len(w) < 1 or len(w) > 10:
            raise ValueError("Количество весов должно быть от 1 до 10")
        self.w = w

    def forward(self, x):
        # Вычисление взвешенной суммы (сумматор)
        if len(x) != len(self.w):
            raise ValueError(f"Длина входного вектора ({len(x)}) должна совпадать с длиной весов ({len(self.w)})")

        # Матричное умножение через цикл
        s = 0
        for i in range(len(self.w)):
            s += self.w[i] * x[i]
        return s

    # Метод для совместимости с примером из ТЗ
    def y(self, x):
        # Вычисление взвешенной суммы (как в примере ТЗ)
        return self.forward(x)

    # Функции активации из урока
    def unit_activation(self, x):
        # Единичная функция активации: y = 1 если s > 0, иначе 0
        s = self.forward(x)
        return 1 if s > 0 else 0

    def binary_activation(self, x, threshold=0.5):
        # Бинарная функция активации: y = 1 если s >= threshold, иначе 0
        s = self.forward(x)
        return 1 if s >= threshold else 0

    def heaviside_activation(self, x):
        # Функция Хевисайда: y = 0 если s < 0, y = 1 если s >= 0
        s = self.forward(x)
        return 0 if s < 0 else 1
```
