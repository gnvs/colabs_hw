1. **Асинхронный таймер:**
   - Напишите программу с использованием `asyncio`, которая в течение определенного времени (например, 10 секунд) выводит сообщение каждую секунду. Используйте `asyncio.sleep` для создания таймера.

```python
import asyncio

async def timer(duration: int):
    """Асинхронный таймер"""
    for i in range(duration):
        print(f"Прошло {i + 1} секунд(ы)")
        await asyncio.sleep(1)

    print(f"Таймер завершен! Прошло {duration} секунд")

await timer(10)
```

2. **Асинхронный калькулятор:**
   - Создайте программу, где пользователь может вводить асинхронные математические операции (например, сложение, умножение) в интерактивном режиме. Реализуйте асинхронную обработку ввода.

```python
import asyncio

async def calculate():
    print("Калькулятор")
    print("Операции: + (сложение), - (вычитание), * (умножение), / (деление)")
    print("Примеры: 2+3, 10-4, 5*6, 15/3")
    print("Для выхода введите 'exit'")

    running = True
    while running:
        try:
            expr = await asyncio.to_thread(input, "\nВыражение: ")

            if expr.lower() == 'exit':
                print("До свидания!")
                await asyncio.sleep(0.05)
                running = False
                break

            if not expr:
                continue

            allowed_chars = set('0123456789+-*/(). ')
            if not all(c in allowed_chars for c in expr):
                print("Ошибка: используйте только числа и операции +, -, *, /")
                continue

            try:
                result = eval(expr)
                print(f"{expr} = {result}")
            except ZeroDivisionError:
                print("Ошибка: деление на ноль!")
            except SyntaxError:
                print("Ошибка: неверный синтаксис. Пример: 2+3")
            except Exception as e:
                print(f"Ошибка: {e}")

        except (asyncio.CancelledError, KeyboardInterrupt):
            print("\nДо свидания!")
            running = False
            break

# Запуск
await calculate()
```

3. **Асинхронные задачи с обработкой ошибок:**
   - Напишите программу, в которой несколько асинхронных задач выполняются параллельно. Включите обработку возможных ошибок в каждой задаче.

```python
import asyncio
import random

async def task(name: str, duration: float, should_fail: bool = False):
    """Простая задача"""
    try:
        print(f"Задача '{name}' началась (будет выполняться {duration}с)")

        if duration < 0:
            raise ValueError("Длительность не может быть отрицательной")

        await asyncio.sleep(duration)

        if should_fail:
            raise Exception(f"Преднамеренная ошибка в задаче '{name}'")

        print(f"Задача '{name}' завершена успешно")
        return {"name": name, "result": f"Результат {name}", "error": None}

    except Exception as e:
        print(f"Задача '{name}' упала с ошибкой: {e}")
        return {"name": name, "result": None, "error": str(e)}

async def main():
    print("Запускаем параллельные задачи...\n")

    # Список задач
    tasks = [
        task("Загрузка данных", 2.0, False),
        task("Обработка изображения", 1.0, True),   # Эта задача упадет
        task("Отправка email", 1.5, False),
        task("Резервное копирование", 0.5, True),  # Эта задача упадет
        task("Генерация отчета", 2.5, False),
    ]

    # Запускаем все параллельно
    results = await asyncio.gather(*tasks)

    print("\n" + "="*50)
    print("ИТОГОВЫЙ ОТЧЕТ:")
    print("="*50)

    for result in results:
        if result["error"]:
            print(f"{result['name']}: ОШИБКА - {result['error']}")
        else:
            print(f"{result['name']}: {result['result']}")

    # Подсчет статистики
    successful = [r for r in results if not r["error"]]
    failed = [r for r in results if r["error"]]

    print(f"\nСтатистика: Успешно: {len(successful)}, Ошибок: {len(failed)}")

# Запуск
await main()
```

4. **Асинхронный генератор случайных чисел:**
   - Реализуйте асинхронный генератор, который асинхронно возвращает случайные числа. Запустите несколько задач для генерации чисел параллельно.

```python
import asyncio
import random

async def random_gen(name: str, count: int):
    """Асинхронный генератор случайных чисел"""
    for i in range(count):
        await asyncio.sleep(1)
        yield f"[{name}] {random.randint(1, 100)}"

async def main():
    # Создаем генераторы
    gen1 = random_gen("A", 3)
    gen2 = random_gen("B", 3)
    gen3 = random_gen("C", 3)

    # Собираем результаты из каждого генератора
    results1 = [i async for i in gen1]
    results2 = [i async for i in gen2]
    results3 = [i async for i in gen3]

    # Запускаем параллельно
    all_results = await asyncio.gather(
        asyncio.create_task(asyncio.sleep(0, results1)),
        asyncio.create_task(asyncio.sleep(0, results2)),
        asyncio.create_task(asyncio.sleep(0, results3))
    )

    # Выводим
    for res in all_results:
        print('\n'.join(res))

# Запуск
await main()
```

5. **Асинхронный поиск файлов:**
   - Напишите программу, которая асинхронно ищет файлы определенного типа в указанном каталоге. Обработайте результаты поиска.

```python
import asyncio
import os
from pathlib import Path

async def find_files(directory: str, extension: str):
    """Асинхронный поиск файлов с заданным расширением"""
    found_files = []

    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(extension):
                file_path = os.path.join(root, file)
                found_files.append(file_path)
                print(f"📁 Найден: {file_path}")
            await asyncio.sleep(0)  # Даем возможность другим задачам выполниться

    return found_files

async def main():
    directory = input("Введите путь к каталогу: ")
    extension = input("Введите расширение (например: .txt, .py, .jpg): ")

    print(f"\nПоиск файлов с расширением {extension} в {directory}...\n")

    files = await find_files(directory, extension)

    print(f"\nНайдено файлов: {len(files)}")

    if files:
        print("\nСписок найденных файлов:")
        for f in files:
            print(f"  • {f}")

# Запуск
await main()
```

6. **Асинхронный кэш:**
   - Создайте простой асинхронный кэш, который позволяет асинхронно добавлять и получать значения по ключу. Проверьте его работу в параллельных задачах.

```python
import asyncio
import time

class AsyncCache:
    """Простой асинхронный кэш"""

    def __init__(self):
        self._cache = {}
        self._lock = asyncio.Lock()

    async def set(self, key, value, ttl=None):
        """Асинхронно добавить значение в кэш"""
        async with self._lock:
            self._cache[key] = {
                'value': value,
                'time': time.time(),
                'ttl': ttl
            }
            print(f"[{key}] = {value} сохранено")

    async def get(self, key):
        """Асинхронно получить значение из кэша"""
        async with self._lock:
            if key not in self._cache:
                print(f"[{key}] не найден")
                return None

            item = self._cache[key]

            # Проверка TTL
            if item['ttl'] and time.time() - item['time'] > item['ttl']:
                del self._cache[key]
                print(f"[{key}] истек (TTL)")
                return None

            print(f"[{key}] = {item['value']} получено")
            return item['value']

async def worker(cache: AsyncCache, key: str, value: int, delay: float):
    """Рабочая задача"""
    await asyncio.sleep(delay)
    await cache.set(key, value)
    result = await cache.get(key)
    return result

async def main():
    cache = AsyncCache()

    # Параллельные задачи
    tasks = [
        worker(cache, "a", 100, 0.5),
        worker(cache, "b", 200, 0.3),
        worker(cache, "c", 300, 0.7),
    ]

    results = await asyncio.gather(*tasks)
    print(f"\nРезультаты: {results}")

# Запуск
await main()
```

7. **Асинхронный счетчик:**
   - Создайте асинхронный счетчик, который можно увеличивать и уменьшать из разных задач параллельно. Обеспечьте синхронный доступ к счетчику.

```python
import asyncio

class AsyncCounter:
    """Асинхронный счетчик с блокировкой"""

    def __init__(self, initial=0):
        self._value = initial
        self._lock = asyncio.Lock()

    async def increment(self, delta=1):
        async with self._lock:
            self._value += delta
            print(f"+{delta} = {self._value}")
            return self._value

    async def decrement(self, delta=1):
        async with self._lock:
            self._value -= delta
            print(f"-{delta} = {self._value}")
            return self._value

    async def get(self):
        async with self._lock:
            return self._value

async def worker(counter: AsyncCounter, name: str, operations: int):
    """Рабочая задача"""
    for i in range(operations):
        if i % 2 == 0:
            await counter.increment(1)
        else:
            await counter.decrement(1)
        await asyncio.sleep(0.1)
    print(f"{name} завершил работу")

async def main():
    counter = AsyncCounter()

    print(f"Старт: {await counter.get()}\n")

    # Запускаем 3 задачи параллельно
    tasks = [
        worker(counter, "A", 4),
        worker(counter, "B", 4),
        worker(counter, "C", 4),
    ]

    await asyncio.gather(*tasks)

    print(f"\nФинальное значение: {await counter.get()}")

# Запуск
await main()
```

8. **Асинхронные задачи с ожиданием завершения:**
    - Напишите программу, где одна асинхронная задача запускает другие асинхронные задачи и ожидает их завершения. Проверьте, что задачи выполняются параллельно.

```python
import asyncio
import random

async def worker(name: str, duration: float):
    """Отдельная задача"""
    print(f"Задача {name} началась (на {duration}с)")
    await asyncio.sleep(duration)
    print(f"Задача {name} завершилась")
    return f"Результат {name}"

async def supervisor():
    """Главная задача, которая запускает и ожидает другие"""
    print("Супервизор: Запускаю дочерние задачи...")

    # Создаем дочерние задачи
    tasks = [
        asyncio.create_task(worker("A", 2)),
        asyncio.create_task(worker("B", 1)),
        asyncio.create_task(worker("C", 3)),
    ]

    print("Супервизор: Ожидаю завершения всех задач...")

    # Ожидаем завершения всех задач
    results = await asyncio.gather(*tasks)

    print(f"Супервизор: Все задачи завершены!")
    print(f"Результаты: {results}")

async def main():
    await supervisor()

# Запуск
await main()
```
