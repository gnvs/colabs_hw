1. **Асинхронный таймер:**
   - Напишите программу с использованием `asyncio`, которая в течение определенного времени (например, 10 секунд) выводит сообщение каждую секунду. Используйте `asyncio.sleep` для создания таймера.

```python
import asyncio

async def print_message(seconds):
    """Выводит сообщение каждую секунду в течение указанного времени."""
    for i in range(1, seconds + 1):
        print(f"Прошло {i} секунд(ы)")
        await asyncio.sleep(1)

async def main():
    duration = 10
    print(f"Программа будет работать {duration} секунд\n")
    await print_message(duration)
    print(f"\nПрограмма завершена через {duration} секунд")

# В Jupyter/IPython просто используйте await
await main()
```

2. **Асинхронный калькулятор:**
   - Создайте программу, где пользователь может вводить асинхронные математические операции (например, сложение, умножение) в интерактивном режиме. Реализуйте асинхронную обработку ввода.

```python
import asyncio
import re
from typing import Tuple, Optional

class AsyncCalculator:
    """Асинхронный калькулятор с интерактивным вводом."""

    def __init__(self):
        self.history = []
        self.operations = {
            '+': lambda a, b: a + b,
            '-': lambda a, b: a - b,
            '*': lambda a, b: a * b,
            '/': lambda a, b: a / b if b != 0 else float('inf'),
            '^': lambda a, b: a ** b,
            '//': lambda a, b: a // b if b != 0 else float('inf'),
            '%': lambda a, b: a % b if b != 0 else float('inf')
        }

    def parse_expression(self, expression: str) -> Optional[Tuple[float, str, float]]:
        """Парсинг математического выражения."""
        expression = expression.strip()

        # Шаблон для поиска: число оператор число
        patterns = [
            (r'^(\d+(?:\.\d+)?)\s*([+\-*/^])\s*(\d+(?:\.\d+)?)$'),  # для + - * / ^
            (r'^(\d+(?:\.\d+)?)\s*(//)\s*(\d+(?:\.\d+)?)$'),        # для целочисленного деления
            (r'^(\d+(?:\.\d+)?)\s*(%)\s*(\d+(?:\.\d+)?)$')          # для остатка
        ]

        for pattern in patterns:
            match = re.match(pattern, expression)
            if match:
                a, op, b = match.groups()
                return (float(a), op, float(b))

        return None

    async def calculate(self, expression: str) -> Tuple[bool, str]:
        """Асинхронное вычисление выражения."""
        # Имитируем асинхронную обработку
        await asyncio.sleep(0.1)

        parsed = self.parse_expression(expression)
        if not parsed:
            return (False, "Ошибка: Неверный формат. Используйте: число оператор число")

        a, op, b = parsed

        if op not in self.operations:
            return (False, f"Ошибка: Оператор '{op}' не поддерживается")

        if op == '/' and b == 0:
            return (False, "Ошибка: Деление на ноль")

        try:
            result = self.operations[op](a, b)
            if result == float('inf'):
                return (False, "Ошибка: Математическая ошибка")

            calculation = f"{a} {op} {b} = {result}"
            self.history.append(calculation)
            return (True, f"Результат: {result}")

        except Exception as e:
            return (False, f"Ошибка: {str(e)}")

    async def run_interactive(self):
        """Запуск интерактивного режима калькулятора."""
        print("=" * 50)
        print("Асинхронный калькулятор")
        print("=" * 50)
        print("Поддерживаемые операции: +, -, *, /, ^, //, %")
        print("Формат ввода: число оператор число (например: 2 + 2)")
        print("Команды: 'history' - показать историю, 'clear' - очистить историю, 'exit' - выход")
        print("=" * 50)

        while True:
            # Асинхронный ввод
            user_input = await asyncio.to_thread(input, "\nВведите выражение: ")

            if user_input.lower() == 'exit':
                print("До свидания!")
                break
            elif user_input.lower() == 'history':
                await self.show_history()
                continue
            elif user_input.lower() == 'clear':
                self.history.clear()
                print("История очищена")
                continue
            elif not user_input:
                continue

            # Асинхронное вычисление
            success, message = await self.calculate(user_input)
            print(message)

    async def show_history(self):
        """Показать историю вычислений."""
        if not self.history:
            print("История пуста")
        else:
            print("\nИстория вычислений:")
            for i, calc in enumerate(self.history, 1):
                print(f"  {i}. {calc}")

# Запуск
calculator = AsyncCalculator()
await calculator.run_interactive()
```

3. **Асинхронные задачи с обработкой ошибок:**
   - Напишите программу, в которой несколько асинхронных задач выполняются параллельно. Включите обработку возможных ошибок в каждой задаче.

```python
import asyncio
import random

async def task(name, duration, fail_chance):
    """
    Простая асинхронная задача.

    Args:
        name: имя задачи
        duration: время выполнения (секунды)
        fail_chance: вероятность ошибки (0-1)
    """
    print(f"Задача '{name}' началась (будет выполняться {duration} сек)")

    try:
        # Имитируем работу
        await asyncio.sleep(duration)

        # Проверяем, не должна ли задача упасть с ошибкой
        if random.random() < fail_chance:
            raise Exception(f"Задача '{name}' упала с ошибкой!")

        # Если всё хорошо
        print(f"Задача '{name}' успешно выполнена!")
        return f"Результат задачи '{name}'"

    except Exception as error:
        print(f"Задача '{name}': {error}")
        return None  # Возвращаем None при ошибке

async def main():
    print("=" * 50)
    print("Запуск асинхронных задач с обработкой ошибок")
    print("=" * 50)

    # Создаём список задач
    # Каждая задача: имя, время выполнения, вероятность ошибки
    tasks = [
        task("Загрузка данных", 2, 0.1),    # 10% ошибки
        task("Обработка файла", 1, 0.3),    # 30% ошибки
        task("Отправка запроса", 3, 0.5),   # 50% ошибки
        task("Сохранение результатов", 1.5, 0.2)  # 20% ошибки
    ]

    # Запускаем все задачи параллельно
    # return_exceptions=True - не прерываем программу при ошибках
    results = await asyncio.gather(*tasks, return_exceptions=True)

    print("\n" + "=" * 50)
    print("Результаты выполнения:")
    print("=" * 50)

    # Выводим результаты
    for i, result in enumerate(results, 1):
        if isinstance(result, Exception):
            print(f"{i}. Ошибка: {result}")
        elif result is None:
            print(f"{i}. Задача не выполнена")
        else:
            print(f"{i}. {result}")

    # Подводим итоги
    successful = sum(1 for r in results if r is not None and not isinstance(r, Exception))
    failed = len(results) - successful
    print(f"\n Итог: Успешно: {successful}, Ошибок: {failed}")

# Запуск
await main()
```

4. **Асинхронный генератор случайных чисел:**
   - Реализуйте асинхронный генератор, который асинхронно возвращает случайные числа. Запустите несколько задач для генерации чисел параллельно.

```python
import asyncio
import random

async def random_number_generator(name, count, delay):
    """
    Асинхронный генератор случайных чисел.
    """
    print(f"Генератор '{name}' запущен. Сгенерирую {count} чисел с задержкой {delay}с")
    results = []

    for i in range(1, count + 1):
        await asyncio.sleep(delay)
        number = random.randint(1, 100)
        print(f"Генератор '{name}': число #{i} = {number}")
        results.append(f"{name}: {number}")

    print(f"Генератор '{name}' завершил работу")
    return results

async def main():
    print("=" * 50)
    print("Асинхронные генераторы случайных чисел")
    print("=" * 50)

    # Запускаем все генераторы параллельно
    results = await asyncio.gather(
        random_number_generator("A", 5, 0.5),
        random_number_generator("B", 3, 1.0),
        random_number_generator("C", 4, 0.8)
    )

    print("\n" + "=" * 50)
    print("Все сгенерированные числа:")
    for gen_results in results:
        for result in gen_results:
            print(f"  {result}")

await main()
```

5. **Асинхронный поиск файлов:**
   - Напишите программу, которая асинхронно ищет файлы определенного типа в указанном каталоге. Обработайте результаты поиска.

```python
import asyncio
import os
from pathlib import Path
from typing import List

async def search_files(directory: str, extension: str, search_id: int):
    """
    Асинхронный поиск файлов с определённым расширением.

    Args:
        directory: каталог для поиска
        extension: расширение файла (например, '.txt', '.py')
        search_id: идентификатор поиска
    """
    print(f"[Поиск {search_id}] Начинаю поиск в {directory}")

    found_files = []

    try:
        # Проверяем существование каталога
        if not os.path.exists(directory):
            raise FileNotFoundError(f"Каталог {directory} не существует")

        # Асинхронно обходим каталог
        for root, dirs, files in os.walk(directory):
            # Имитируем асинхронность при обработке
            await asyncio.sleep(0.01)

            for file in files:
                if file.endswith(extension):
                    full_path = os.path.join(root, file)
                    found_files.append(full_path)
                    print(f"[Поиск {search_id}] Найден: {file}")

        return found_files

    except Exception as e:
        print(f"[Поиск {search_id}] Ошибка: {e}")
        return []

async def main():
    print("=" * 60)
    print("Асинхронный поиск файлов")
    print("=" * 60)

    # Запускаем несколько поисков параллельно
    results = await asyncio.gather(
        search_files(".", ".py", 1),      # Поиск Python файлов
        search_files(".", ".txt", 2),     # Поиск текстовых файлов
        search_files(".", ".md", 3)       # Поиск Markdown файлов
    )

    print("\n" + "=" * 60)
    print("Результаты поиска:")
    print("=" * 60)

    extensions = [".py", ".txt", ".md"]
    for i, files in enumerate(results):
        print(f"\n📁 Расширение {extensions[i]}: найдено {len(files)} файлов")
        for file in files[:5]:  # Показываем первые 5
            print(f"   📄 {file}")
        if len(files) > 5:
            print(f"   ... и ещё {len(files) - 5} файлов")

await main()
```

6. **Асинхронный кэш:**
   - Создайте простой асинхронный кэш, который позволяет асинхронно добавлять и получать значения по ключу. Проверьте его работу в параллельных задачах.

```python
import asyncio
import time
from datetime import datetime, timedelta

class AsyncCache:
    """Простой асинхронный кэш."""

    def __init__(self):
        self.cache = {}  # Хранилище {ключ: значение}
        self.ttl = {}    # Время жизни {ключ: время_истечения}

    async def set(self, key, value, ttl_seconds=60):
        """
        Асинхронно добавляет значение в кэш.

        Args:
            key: ключ
            value: значение
            ttl_seconds: время жизни в секундах
        """
        # Имитируем асинхронную операцию записи
        await asyncio.sleep(0.01)

        self.cache[key] = value
        if ttl_seconds > 0:
            expire_time = datetime.now() + timedelta(seconds=ttl_seconds)
            self.ttl[key] = expire_time

        print(f"[КЭШ] Добавлен: {key} = {value} (TTL: {ttl_seconds}с)")

    async def get(self, key):
        """
        Асинхронно получает значение из кэша.

        Args:
            key: ключ

        Returns:
            значение или None, если ключ не найден
        """
        # Имитируем асинхронную операцию чтения
        await asyncio.sleep(0.01)

        # Проверяем время жизни
        if key in self.ttl:
            if datetime.now() > self.ttl[key]:
                # Ключ истёк
                del self.cache[key]
                del self.ttl[key]
                print(f"[КЭШ] Ключ {key} истёк")
                return None

        value = self.cache.get(key)
        if value is not None:
            print(f"[КЭШ] Получен: {key} = {value}")
        else:
            print(f"[КЭШ] Ключ {key} не найден")

        return value

    async def delete(self, key):
        """Асинхронно удаляет значение из кэша."""
        await asyncio.sleep(0.01)

        if key in self.cache:
            del self.cache[key]
            if key in self.ttl:
                del self.ttl[key]
            print(f"[КЭШ] Удалён: {key}")
            return True
        return False

    async def clear(self):
        """Очищает весь кэш."""
        await asyncio.sleep(0.01)
        self.cache.clear()
        self.ttl.clear()
        print(f"🧹 [КЭШ] Кэш полностью очищен")

    async def stats(self):
        """Возвращает статистику кэша."""
        await asyncio.sleep(0.01)
        return {
            'size': len(self.cache),
            'keys': list(self.cache.keys()),
            'expired': sum(1 for k in self.ttl if datetime.now() > self.ttl[k])
        }

async def test_cache():
    """Тестирование асинхронного кэша."""

    cache = AsyncCache()

    print("=" * 60)
    print("Тестирование асинхронного кэша")
    print("=" * 60)

    # Добавляем значения
    await cache.set("user:1", "Alice", ttl_seconds=5)
    await cache.set("user:2", "Bob", ttl_seconds=10)
    await cache.set("config", "settings", ttl_seconds=3)

    print("\n--- Получение значений ---")
    await cache.get("user:1")
    await cache.get("user:2")
    await cache.get("unknown")

    print("\n--- Ждём истечения TTL (3 секунды) ---")
    await asyncio.sleep(3)

    await cache.get("config")  # Должен истечь
    await cache.get("user:1")  # Ещё действителен

    stats = await cache.stats()
    print(f"\nСтатистика: {stats}")

    await cache.delete("user:1")
    await cache.clear()

await test_cache()
```

7. **Асинхронный счетчик:**
   - Создайте асинхронный счетчик, который можно увеличивать и уменьшать из разных задач параллельно. Обеспечьте синхронный доступ к счетчику.

```python
import asyncio
import random

class AsyncCounter:
    """Асинхронный счетчик с безопасным доступом из разных задач."""

    def __init__(self, initial_value=0):
        self.value = initial_value
        self._lock = asyncio.Lock()  # Блокировка для синхронизации
        self._history = []  # История изменений для демонстрации

    async def increment(self, amount=1, task_name="unknown"):
        """
        Увеличивает счетчик на указанное значение.

        Args:
            amount: значение для увеличения
            task_name: имя задачи (для логирования)
        """
        async with self._lock:  # Захватываем блокировку
            # Имитируем асинхронную операцию (например, чтение из БД)
            await asyncio.sleep(0.1)

            old_value = self.value
            self.value += amount

            # Логируем изменение
            change = f"+{amount}"
            self._history.append({
                'task': task_name,
                'change': change,
                'old': old_value,
                'new': self.value
            })

            print(f"[{task_name}] {old_value} {change} = {self.value}")

    async def decrement(self, amount=1, task_name="unknown"):
        """
        Уменьшает счетчик на указанное значение.

        Args:
            amount: значение для уменьшения
            task_name: имя задачи (для логирования)
        """
        async with self._lock:  # Захватываем блокировку
            # Имитируем асинхронную операцию
            await asyncio.sleep(0.1)

            old_value = self.value
            self.value -= amount

            # Логируем изменение
            change = f"-{amount}"
            self._history.append({
                'task': task_name,
                'change': change,
                'old': old_value,
                'new': self.value
            })

            print(f"[{task_name}] {old_value} {change} = {self.value}")

    async def get_value(self, task_name="unknown"):
        """
        Получает текущее значение счетчика.

        Args:
            task_name: имя задачи (для логирования)
        """
        async with self._lock:
            await asyncio.sleep(0.05)
            print(f"[{task_name}] Текущее значение: {self.value}")
            return self.value

    async def reset(self, task_name="unknown"):
        """Сбрасывает счетчик в 0."""
        async with self._lock:
            await asyncio.sleep(0.1)
            old_value = self.value
            self.value = 0
            self._history.append({
                'task': task_name,
                'change': 'reset',
                'old': old_value,
                'new': 0
            })
            print(f"[{task_name}] Сброс: {old_value} -> 0")

    def show_history(self):
        """Показывает историю изменений."""
        print("\n" + "="*50)
        print("История изменений счетчика:")
        for i, entry in enumerate(self._history, 1):
            print(f"  {i}. {entry['task']}: {entry['old']} → {entry['new']} ({entry['change']})")

async def worker(task_id, counter, operations=5):
    """
    Задача-воркер, которая выполняет операции со счетчиком.

    Args:
        task_id: идентификатор задачи
        counter: объект AsyncCounter
        operations: количество операций
    """
    task_name = f"Задача-{task_id}"

    for i in range(operations):
        # Выбираем случайную операцию
        operation = random.choice(['increment', 'decrement', 'get'])

        if operation == 'increment':
            amount = random.randint(1, 5)
            await counter.increment(amount, task_name)
        elif operation == 'decrement':
            amount = random.randint(1, 5)
            await counter.decrement(amount, task_name)
        else:  # get
            await counter.get_value(task_name)

        # Небольшая задержка между операциями
        await asyncio.sleep(random.uniform(0.2, 0.5))

    print(f"[{task_name}] Завершила работу")

async def main():
    print("=" * 60)
    print("Асинхронный счетчик с синхронизацией доступа")
    print("=" * 60)
    print("\nЗапускаем 5 параллельных задач, которые работают со счетчиком")
    print("Благодаря блокировке (Lock) все операции безопасны\n")

    # Создаем счетчик
    counter = AsyncCounter(initial_value=100)

    # Создаем несколько параллельных задач
    tasks = []
    for i in range(1, 5):
        task = asyncio.create_task(worker(i, counter, operations=4))
        tasks.append(task)

    # Ждем завершения всех задач
    await asyncio.gather(*tasks)

    # Показываем финальное значение и историю
    final_value = await counter.get_value("Main")
    print(f"\nФинальное значение счетчика: {final_value}")
    counter.show_history()

await main()
```

8. **Асинхронные задачи с ожиданием завершения:**
    - Напишите программу, где одна асинхронная задача запускает другие асинхронные задачи и ожидает их завершения. Проверьте, что задачи выполняются параллельно.

```python
import asyncio
import random

async def worker_task(task_id, duration):
    """
    Дочерняя задача (воркер).

    Args:
        task_id: идентификатор задачи
        duration: продолжительность выполнения
    """
    print(f"Задача-{task_id}: запущена (будет работать {duration} сек)")

    # Имитируем полезную работу
    await asyncio.sleep(duration)

    # Имитируем случайный результат
    result = random.randint(1, 100)
    print(f"Задача-{task_id}: завершена (результат: {result})")

    return result

async def main_task():
    """
    Главная задача, которая запускает дочерние задачи и ждёт их.
    """
    print("Главная задача: запускаю дочерние задачи")
    print("-" * 50)

    # Создаём список дочерних задач
    tasks = [
        worker_task(1, 2),
        worker_task(2, 3),
        worker_task(3, 1),
        worker_task(4, 2.5)
    ]

    # Запускаем все задачи параллельно и ждём их завершения
    results = await asyncio.gather(*tasks)

    print("-" * 50)
    print("Главная задача: все дочерние задачи завершены!")
    print(f"Результаты: {results}")
    print(f"Сумма результатов: {sum(results)}")

    return results

# Запуск программы
print("=" * 50)
print("Запуск главной задачи...")
print("=" * 50)

await main_task()

print("\nПрограмма завершена!")
```
