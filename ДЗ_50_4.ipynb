```python
import os
import torch
from IPython.display import Audio, display
from TTS.api import TTS

# Модель

model_name = "tts_models/multilingual/multi-dataset/xtts_v2"

print("Загружаем модель:", model_name)

tts = TTS(model_name)

print("Модель загружена")


# Скачиваем образец голоса

speaker_file = "speaker.wav"

if not os.path.exists(speaker_file):
    print("Скачиваем образец голоса...")

    !wget -q \
    https://github.com/coqui-ai/TTS/raw/dev/tests/data/ljspeech/wavs/LJ001-0001.wav \
    -O speaker.wav

if os.path.exists(speaker_file):
    print("✓ Образец голоса:", speaker_file)
    print("Размер:", os.path.getsize(speaker_file), "байт")
else:
    raise FileNotFoundError("Не удалось скачать speaker.wav")


# Тексты для синтеза

russian_text = """
Здравствуйте! Это пример синтеза речи на русском языке.
Современные технологии позволяют преобразовывать обычный текст
в естественно звучащую человеческую речь.
"""

english_text = """
Hello! This is an example of speech synthesis in English.
Modern technologies can convert ordinary text
into natural sounding human speech.
"""


# Русский язык

russian_file = "speech_russian.wav"

print("\nГенерируем русскую речь...")

tts.tts_to_file(
    text=russian_text,
    speaker_wav=speaker_file,
    language="ru",
    file_path=russian_file
)

print("Русская речь сохранена:", russian_file)

# Английский язык

english_file = "speech_english.wav"

print("\nГенерируем английскую речь...")

tts.tts_to_file(
    text=english_text,
    speaker_wav=speaker_file,
    language="en",
    file_path=english_file
)

print("Английская речь сохранена:", english_file)


# Прослушивание русского варианта

print("\n" + "=" * 50)
print("РУССКИЙ ЯЗЫК")
print("=" * 50)

display(Audio(russian_file))


# Прослушивание английского варианта

print("\n" + "=" * 50)
print("АНГЛИЙСКИЙ ЯЗЫК")
print("=" * 50)

display(Audio(english_file))


# Информация о результатах

print("\n" + "=" * 50)
print("РЕЗУЛЬТАТЫ")
print("=" * 50)

print("Русский файл :", russian_file)
print("Размер        :", os.path.getsize(russian_file), "байт")

print()

print("Английский файл:", english_file)
print("Размер         :", os.path.getsize(english_file), "байт")

print("\nСинтез речи завершён!")
```
