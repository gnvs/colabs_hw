```python
# Mozart

# Установка необходимых библиотек

# FluidSynth нужен для преобразования MIDI в WAV-аудиофайл
!apt-get -qq update
!apt-get -qq install -y fluidsynth fluid-soundfont-gm

# Устанавливаем библиотеки для работы с MIDI.
%pip install -q pretty_midi pyfluidsynth

# Импорт библиотек

import collections
import os
import random
import subprocess
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Библиотека для чтения и создания MIDI-файлов
import pretty_midi

# TensorFlow для построения и обучения нейросети
import tensorflow as tf

# Для вывода аудиоплеера в Colab
from IPython.display import display, Audio

# Фиксируем случайность для воспроизводимости эксперимента
SEED = 42

os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)


print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))

# ===
# Длина последовательности нот, которую видит RNN
SEQ_LENGTH = 50

# MIDI поддерживает pitch от 0 до 127
VOCAB_SIZE = 128

# Размер пакета при обучении
BATCH_SIZE = 64

# Максимальное число эпох.
# EarlyStopping обычно завершит обучение раньше.
EPOCHS = 60

# Число новых нот, которые сгенерирует модель
NUM_GENERATED_NOTES = 300

# температуру можно менять без повторного обучения модели
TEMPERATURE = 1.0

# Ссылка на датасет из задания
DATASET_URL = (
    "https://storage.yandexcloud.net/academy.ai/"
    "classical-music-midi.zip"
)

# Файлы результата
BEST_WEIGHTS = "/content/chopin_best.weights.h5"
MIDI_OUT = "/content/chopin_from_mozart.mid"
WAV_OUT = "/content/chopin_from_mozart.wav"

# Порядок трёх признаков ноты.
# Он должен быть одинаковым во всех частях программы.
KEY_ORDER = [
    "pitch",
    "step",
    "duration",
]

# ===
# tf.keras.utils.get_file скачивает архив и распаковывает его.
archive_path = tf.keras.utils.get_file(
    fname="classical-music-midi.zip",
    origin=DATASET_URL,
    extract=True,
    cache_dir="/content",
    cache_subdir="classical_music_dataset",
)

DATA_ROOT = Path(
    archive_path
).parent


# Ищем .mid и .midi независимо от регистра расширения
all_midi_files = sorted(
    [
        path
        for path in DATA_ROOT.rglob("*")
        if (
            path.is_file()
            and path.suffix.lower() in {
                ".mid",
                ".midi",
            }
        )
    ]
)


print(
    "Всего MIDI-файлов найдено:",
    len(all_midi_files),
)


# Покажем несколько путей, чтобы было видно структуру архива
for path in all_midi_files[:15]:
    print(path)

# ==============================================================================
def find_composer_files(
    midi_files,
    aliases,
):
    """
    Возвращает MIDI-файлы композитора.

    Проверяем полный путь в нижнем регистре, поэтому подходит как структура с папкой Composer, так и архив, где имя композитора находится в имени файла.
    """
    result = []

    for path in midi_files:
        full_name = str(
            path
        ).lower()

        if any(
            alias.lower() in full_name
            for alias in aliases
        ):
            result.append(
                path
            )

    return result

# Только Шопен
chopin_files = find_composer_files(
    all_midi_files,
    aliases=[
        "chopin",
        "chpn",
    ],
)

# MOZART начальной последовательности
mozart_files = find_composer_files(
    all_midi_files,
    aliases=[
        "mozart",
    ],
)


print(
    "Произведений Шопена:",
    len(chopin_files),
)

print(
    "Произведений Моцарта:",
    len(mozart_files),
)


# Если структура архива неожиданно изменилась, сразу получим ошибку.
if not chopin_files:
    raise RuntimeError(
        "Не найдены MIDI Шопена. "
        "Посмотрите напечатанную выше структуру архива."
    )

if not mozart_files:
    raise RuntimeError(
        "Не найдены MIDI Моцарта. "
        "Посмотрите напечатанную выше структуру архива."
    )


print("\nПервые MIDI Шопена:")

for path in chopin_files[:5]:
    print(path)


print("\nПервые MIDI Моцарта:")

for path in mozart_files[:5]:
    print(path)

# ===
def midi_to_notes(
    midi_file,
):
    """
    Преобразует один MIDI-файл в DataFrame:
    pitch | step | duration | start | end
    """

    pm = pretty_midi.PrettyMIDI(
        str(midi_file)
    )


    # Собираем ноты со всех не ударных дорожек.
    # Для классических фортепианных MIDI это позволяет сохранить максимум музыкальной информации.
    midi_notes = []

    for instrument in pm.instruments:

        if instrument.is_drum:
            continue

        midi_notes.extend(
            instrument.notes
        )


    if not midi_notes:
        return pd.DataFrame(
            columns=[
                "pitch",
                "step",
                "duration",
                "start",
                "end",
            ]
        )


    # Сортируем ноты по началу звучания.
    # Вроде ноты аккорда могут иметь одинаковый start, тогда для следующих нот step будет равен 0.
    sorted_notes = sorted(
        midi_notes,
        key=lambda note: (
            note.start,
            note.pitch,
        ),
    )


    notes = collections.defaultdict(
        list
    )

    # Для первой ноты предыдущим стартом считаем 0
    prev_start = 0.0


    for note in sorted_notes:

        start = float(
            note.start
        )

        end = float(
            note.end
        )

        duration = max(
            0.0,
            end - start,
        )

        step = max(
            0.0,
            start - prev_start,
        )


        notes["pitch"].append(
            int(note.pitch)
        )

        notes["step"].append(
            step
        )

        notes["duration"].append(
            duration
        )

        notes["start"].append(
            start
        )

        notes["end"].append(
            end
        )


        prev_start = start


    return pd.DataFrame(
        notes
    )

# ===
chopin_note_tables = []

bad_chopin_files = []


for midi_file in chopin_files:

    try:
        notes = midi_to_notes(
            midi_file
        )

        # Очень короткие композиции не подходят для создания последовательностей длины SEQ_LENGTH.
        if len(notes) > SEQ_LENGTH:
            chopin_note_tables.append(
                notes
            )

    except Exception as error:

        bad_chopin_files.append(
            (
                midi_file,
                str(error),
            )
        )


# Для статистики объединяем произведения.
# Для создания последовательностей ниже границы произведений они будут сохранены.
all_chopin_notes = pd.concat(
    chopin_note_tables,
    ignore_index=True,
)


print(
    "Успешно прочитано произведений Шопена:",
    len(chopin_note_tables),
)

print(
    "Всего нот Шопена:",
    len(all_chopin_notes),
)

print(
    "Ошибочных MIDI:",
    len(bad_chopin_files),
)


display(
    all_chopin_notes.head(10)
)


# Дополнительная проверка: обучающие данные должны существовать.
if len(all_chopin_notes) <= SEQ_LENGTH:
    raise RuntimeError(
        "Недостаточно нот Шопена для обучения."
    )

# ===
fig, axes = plt.subplots(
    1,
    3,
    figsize=(16, 4),
)


axes[0].hist(
    all_chopin_notes["pitch"],
    bins=40,
)

axes[0].set_title("Pitch")
axes[0].set_xlabel("MIDI pitch")


# Для step и duration отбрасываем только очень длинный хвост на графике, сами данные при этом НЕ удаляем.
step_max = np.percentile(
    all_chopin_notes["step"],
    99,
)

duration_max = np.percentile(
    all_chopin_notes["duration"],
    99,
)


axes[1].hist(
    all_chopin_notes["step"].clip(
        upper=step_max
    ),
    bins=40,
)

axes[1].set_title("Step")
axes[1].set_xlabel("Секунды")


axes[2].hist(
    all_chopin_notes["duration"].clip(
        upper=duration_max
    ),
    bins=40,
)

axes[2].set_title("Duration")
axes[2].set_xlabel("Секунды")


plt.tight_layout()
plt.show()


# Эти границы потом используем при генерации,
# чтобы единичное неудачное предсказание не создало паузу или ноту длиной в десятки секунд.
MAX_STEP = float(
    np.percentile(
        all_chopin_notes["step"],
        99.5,
    )
)

MAX_DURATION = float(
    np.percentile(
        all_chopin_notes["duration"],
        99.5,
    )
)


print(
    "MAX_STEP:",
    MAX_STEP,
)

print(
    "MAX_DURATION:",
    MAX_DURATION,
)

# ===
def create_sequences(
    dataset,
    seq_length,
    vocab_size=VOCAB_SIZE,
):
    """
    Создаёт пары:
        последовательность нот -> следующая нота
    Вход:
        (seq_length, 3)
    Выход:
        {
            pitch,
            step,
            duration
        }
    """

    # +1 — последняя нота станет правильным ответом
    sequence_length = (
        seq_length + 1
    )


    windows = dataset.window(
        sequence_length,
        shift=1,
        stride=1,
        drop_remainder=True,
    )


    # Dataset из маленьких Dataset превращаем в Dataset обычных тензоров
    sequences = windows.flat_map(
        lambda window:
            window.batch(
                sequence_length,
                drop_remainder=True,
            )
    )


    def split_labels(
        sequence,
    ):
        """
        Первые SEQ_LENGTH нот -> X.
        Последняя нота -> y.
        """

        inputs = sequence[:-1]

        label = sequence[-1]


        # pitch — дискретный номер 0..127.
        # На вход модели его нормализуем в диапазон 0..1.
        # step и duration оставляем в секундах.
        inputs = inputs / tf.constant(
            [
                float(vocab_size),
                1.0,
                1.0,
            ],
            dtype=tf.float32,
        )


        labels = {
            "pitch": label[0],
            "step": label[1],
            "duration": label[2],
        }


        return (
            inputs,
            labels,
        )


    return sequences.map(
        split_labels,
        num_parallel_calls=tf.data.AUTOTUNE,
    )

# ===
# Общий Dataset будет собран из отдельных произведений.
sequence_ds = None

total_examples = 0


for notes in chopin_note_tables:

    # pitch, step, duration именно в таком порядке
    note_array = np.stack(
        [
            notes[key].to_numpy(
                dtype=np.float32
            )
            for key in KEY_ORDER
        ],
        axis=1,
    )


    piece_ds = (
        tf.data.Dataset
        .from_tensor_slices(
            note_array
        )
    )


    piece_seq_ds = create_sequences(
        piece_ds,
        SEQ_LENGTH,
        VOCAB_SIZE,
    )


    # Склеиваем уже готовые последовательности, а не сами произведения.
    if sequence_ds is None:
        sequence_ds = piece_seq_ds
    else:
        sequence_ds = sequence_ds.concatenate(
            piece_seq_ds
        )


    total_examples += max(
        0,
        len(note_array)
        - SEQ_LENGTH,
    )


print(
    "Обучающих последовательностей:",
    total_examples,
)


if total_examples < 100:
    raise RuntimeError(
        "Слишком мало последовательностей для обучения."
    )

# ===
# Перемешиваем последовательности один раз, чтобы train и validation получили примеры из разных произведений Шопена.
split_ds = sequence_ds.shuffle(
    buffer_size=min(
        total_examples,
        100000,
    ),
    seed=SEED,
    reshuffle_each_iteration=False,
)


val_count = max(
    1,
    int(
        total_examples * 0.10
    ),
)

train_count = (
    total_examples
    - val_count
)


val_ds = (
    split_ds
    .take(val_count)
    .batch(
        BATCH_SIZE
    )
    .prefetch(
        tf.data.AUTOTUNE
    )
)


train_ds = (
    split_ds
    .skip(val_count)
    .shuffle(
        buffer_size=min(
            train_count,
            100000,
        ),
        seed=SEED,
        reshuffle_each_iteration=True,
    )
    .batch(
        BATCH_SIZE
    )
    .prefetch(
        tf.data.AUTOTUNE
    )
)


print(
    "TRAIN последовательностей:",
    train_count,
)

print(
    "VALIDATION последовательностей:",
    val_count,
)


# Показываем формы одного пакета
for x_batch, y_batch in train_ds.take(1):

    print(
        "X:",
        x_batch.shape,
    )

    print(
        "pitch:",
        y_batch["pitch"].shape,
    )

    print(
        "step:",
        y_batch["step"].shape,
    )

    print(
        "duration:",
        y_batch["duration"].shape,
    )

# ===
def mse_with_positive_pressure(
    y_true,
    y_pred,
):
    """
    MSE + штраф за отрицательный прогноз.
    Это помогает step и duration оставаться >= 0.
    """

    mse = tf.square(
        y_true - y_pred
    )

    positive_pressure = (
        10.0
        * tf.maximum(
            -y_pred,
            0.0,
        )
    )

    return tf.reduce_mean(
        mse
        + positive_pressure
    )

# ===
# Вход: последовательность из SEQ_LENGTH нот, каждая нота содержит 3 признака.
inputs = tf.keras.Input(
    shape=(
        SEQ_LENGTH,
        3,
    ),
    name="notes",
)


# LSTM хорошо подходит для последовательностей, потому что учитывает порядок предыдущих нот.
x = tf.keras.layers.LSTM(
    256,
    name="lstm",
)(
    inputs
)


x = tf.keras.layers.Dropout(
    0.20,
)(
    x
)


# Небольшой общий Dense-блок перед тремя выходами
x = tf.keras.layers.Dense(
    128,
    activation="relu",
)(
    x
)

# 3 модели
outputs = {
    # 128 логитов — по одному для каждого MIDI pitch
    "pitch": tf.keras.layers.Dense(
        VOCAB_SIZE,
        name="pitch",
    )(
        x
    ),

    # Временной шаг до следующей ноты
    "step": tf.keras.layers.Dense(
        1,
        name="step",
    )(
        x
    ),

    # Длительность следующей ноты
    "duration": tf.keras.layers.Dense(
        1,
        name="duration",
    )(
        x
    ),
}


model = tf.keras.Model(
    inputs,
    outputs,
)


loss = {
    "pitch":
        tf.keras.losses.SparseCategoricalCrossentropy(
            from_logits=True
        ),

    "step":
        mse_with_positive_pressure,

    "duration":
        mse_with_positive_pressure,
}


optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.003
)


# Pitch-loss обычно численно намного больше двух MSE.
# Поэтому уменьшаем его вес, чтобы модель не игнорировала step и duration.
model.compile(
    optimizer=optimizer,
    loss=loss,
    loss_weights={
        "pitch": 0.05,
        "step": 1.0,
        "duration": 1.0,
    },
)


model.summary()

# ===
callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        BEST_WEIGHTS,
        monitor="val_loss",
        mode="min",
        save_best_only=True,
        save_weights_only=True,
        verbose=1,
    ),

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        mode="min",
        patience=8,
        restore_best_weights=True,
        verbose=1,
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        mode="min",
        factor=0.5,
        patience=4,
        min_lr=1e-5,
        verbose=1,
    ),
]


history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1,
)


# Загружаем лучшие веса
model.load_weights(
    BEST_WEIGHTS
)

# ===
plt.figure(
    figsize=(13, 5)
)

plt.plot(
    history.history["loss"],
    label="train loss",
)

plt.plot(
    history.history["val_loss"],
    label="validation loss",
)

plt.xlabel("Эпоха")
plt.ylabel("Loss")
plt.title(
    "Обучение модели на произведениях Шопена"
)

plt.grid()
plt.legend()
plt.show()

# ===
mozart_seed_notes = None
mozart_seed_file = None


# Ищем первый корректный MIDI Моцарта, содержащий достаточно нот для seed.
for midi_file in mozart_files:

    try:
        notes = midi_to_notes(
            midi_file
        )

        if len(notes) >= SEQ_LENGTH:
            mozart_seed_notes = notes
            mozart_seed_file = midi_file
            break

    except Exception:
        continue


if mozart_seed_notes is None:
    raise RuntimeError(
        "Не удалось найти достаточно длинное "
        "произведение Моцарта для seed."
    )


print(
    "Начальная последовательность взята из:"
)

print(
    mozart_seed_file
)

print(
    "Нот в произведении:",
    len(mozart_seed_notes),
)


display(
    mozart_seed_notes.head(10)
)

# ===
def predict_next_note(
    notes,
    model,
    temperature=1.0,
):
    """
    Генерирует следующую ноту:

        pitch, step, duration

    temperature управляет случайностью выбора pitch.
    """

    if temperature <= 0:
        raise ValueError(
            "temperature должна быть > 0"
        )


    # Добавляем batch-размерность: (SEQ_LENGTH, 3) -> (1, SEQ_LENGTH, 3)
    inputs = tf.expand_dims(
        notes,
        axis=0,
    )


    # Получаем три прогноза
    predictions = model(
        inputs,
        training=False,
    )


    pitch_logits = (
        predictions["pitch"][0]
        / temperature
    )

    # Не берём просто argmax. Сэмплируем ноту из вероятностного распределения.
    pitch = tf.random.categorical(
        tf.expand_dims(
            pitch_logits,
            axis=0,
        ),
        num_samples=1,
    )[0, 0]


    step = tf.squeeze(
        predictions["step"]
    )

    duration = tf.squeeze(
        predictions["duration"]
    )


    # step и duration физически не могут быть отрицательными.
    step = float(
        tf.maximum(
            0.0,
            step,
        ).numpy()
    )

    duration = float(
        tf.maximum(
            0.0,
            duration,
        ).numpy()
    )


    # Защита от предсказания.
    # Верхние границы получены только из нот Шопена.
    step = float(
        np.clip(
            step,
            0.0,
            MAX_STEP,
        )
    )

    duration = float(
        np.clip(
            duration,
            0.03,
            MAX_DURATION,
        )
    )


    return (
        int(
            pitch.numpy()
        ),
        step,
        duration,
    )

# ===
# Берём первые SEQ_LENGTH нот выбранного произведения Моцарта
seed_raw = np.stack(
    [
        mozart_seed_notes[key]
        .iloc[:SEQ_LENGTH]
        .to_numpy(
            dtype=np.float32
        )
        for key in KEY_ORDER
    ],
    axis=1,
)


# Нормализуем pitch точно так же, как в обучающем Dataset.
input_notes = (
    seed_raw
    / np.array(
        [
            VOCAB_SIZE,
            1.0,
            1.0,
        ],
        dtype=np.float32,
    )
)


print(
    "Форма Mozart seed:",
    input_notes.shape,
)

print(
    "TEMPERATURE:",
    TEMPERATURE,
)

# ===
generated_notes = []

prev_start = 0.0


for note_number in range(
    NUM_GENERATED_NOTES
):

    pitch, step, duration = predict_next_note(
        input_notes,
        model,
        temperature=TEMPERATURE,
    )


    # Время начала и конца новой ноты
    start = (
        prev_start
        + step
    )

    end = (
        start
        + duration
    )


    # Сохраняем ноту исходном виде
    generated_notes.append(
        (
            pitch,
            step,
            duration,
            start,
            end,
        )
    )


    # модель обучалась на нормализованном pitch,
    # поэтому обратно в окно добавляем pitch / 128.
    generated_input_note = np.array(
        [
            pitch / VOCAB_SIZE,
            step,
            duration,
        ],
        dtype=np.float32,
    )


    # Удаляем самую старую ноту
    input_notes = input_notes[
        1:
    ]


    # И добавляем новую в конец окна
    input_notes = np.concatenate(
        [
            input_notes,
            generated_input_note[
                None,
                :
            ],
        ],
        axis=0,
    )


    prev_start = start


generated_notes = pd.DataFrame(
    generated_notes,
    columns=[
        "pitch",
        "step",
        "duration",
        "start",
        "end",
    ],
)


print(
    "Сгенерировано нот:",
    len(generated_notes),
)


display(
    generated_notes.head(20)
)

# ===
generated_preview = (
    generated_notes
    .head(30)
    .copy()
)


generated_preview[
    "note_name"
] = generated_preview[
    "pitch"
].apply(
    pretty_midi.note_number_to_name
)


display(
    generated_preview[
        [
            "note_name",
            "pitch",
            "step",
            "duration",
            "start",
            "end",
        ]
    ]
)

# ===
def plot_piano_roll(
    notes,
    count=None,
):
    """
    Простая визуализация нот по времени.
    """

    if count is None:
        count = len(notes)

    subset = notes.iloc[
        :count
    ]


    plt.figure(
        figsize=(16, 5)
    )


    for _, note in subset.iterrows():

        plt.plot(
            [
                note["start"],
                note["end"],
            ],
            [
                note["pitch"],
                note["pitch"],
            ],
            linewidth=2,
        )


    plt.xlabel("Время, сек.")
    plt.ylabel("MIDI pitch")

    plt.title(
        f"Сгенерированное произведение, "
        f"temperature={TEMPERATURE}"
    )

    plt.grid()
    plt.show()


plot_piano_roll(
    generated_notes
)

# ===
def notes_to_midi(
    notes,
    out_file,
    instrument_name="Acoustic Grand Piano",
    velocity=90,
):
    """
    Создаёт PrettyMIDI и записывает его в файл.
    """

    pm = pretty_midi.PrettyMIDI()


    instrument = pretty_midi.Instrument(
        program=(
            pretty_midi
            .instrument_name_to_program(
                instrument_name
            )
        )
    )


    for _, row in notes.iterrows():

        midi_note = pretty_midi.Note(
            velocity=velocity,
            pitch=int(
                row["pitch"]
            ),
            start=float(
                row["start"]
            ),
            end=float(
                row["end"]
            ),
        )


        instrument.notes.append(
            midi_note
        )


    pm.instruments.append(
        instrument
    )


    pm.write(
        str(out_file)
    )


    return pm


generated_pm = notes_to_midi(
    generated_notes,
    MIDI_OUT,
)


print(
    "MIDI создан:",
    MIDI_OUT,
)

print(
    "Размер MIDI:",
    Path(MIDI_OUT).stat().st_size,
    "байт",
)

# ===
# Ищем установленный General MIDI SoundFont
soundfont_candidates = (
    list(
        Path(
            "/usr/share/sounds/sf2"
        ).glob(
            "*.sf2"
        )
    )
    + list(
        Path(
            "/usr/share/sounds/sf3"
        ).glob(
            "*.sf3"
        )
    )
)


if not soundfont_candidates:
    raise RuntimeError(
        "SoundFont не найден. "
        "Проверьте выполнение первой ячейки установки."
    )


SOUNDFONT = soundfont_candidates[
    0
]


print(
    "Используем SoundFont:",
    SOUNDFONT,
)


# Преобразуем MIDI -> WAV.
# -ni = без интерактивного режима
subprocess.run(
    [
        "fluidsynth",
        "-ni",
        str(SOUNDFONT),
        MIDI_OUT,
        "-F",
        WAV_OUT,
        "-r",
        "44100",
    ],
    check=True,
)


print(
    "WAV создан:",
    WAV_OUT,
)

print(
    "Размер WAV:",
    Path(WAV_OUT).stat().st_size,
    "байт",
)

# ===
# Аудиоплеер в Colab-e
display.display(
    display.Audio(
        WAV_OUT
    )
)

# ===
from google.colab import files


# MIDI — ноты сгенерированной композиции
files.download(
    MIDI_OUT
)


# WAV — готовый аудиофайл
files.download(
    WAV_OUT
)
```
